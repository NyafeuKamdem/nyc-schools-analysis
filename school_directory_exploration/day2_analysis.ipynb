## **1. Labraries**

```python
# Importing the necessary labraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## **2. Data Loading and cleaning**

**2.1.Data Loading**

```python
df = pd.read_csv("high-school-directory-day2.csv")
pd.set_option("display.max_columns", None) # None = Show all columns
df.head(3)
```

**2.2. Clean the column names**
- (make lowercase, replace spaces with _, remove special characters)

```python
col_lower = df.columns.str.lower()                                  # Lowercase all column names
col_with_ = col_lower.str.replace(" ", "_")                         # Replace spaces with underscores
df.columns = col_with_.str.replace(r"[^\w_]", "", regex = True)     # Remove all non-word characters
df.head(3)
```

```python
df.info(verbose = True, show_counts = True)                         # display all columns' name and general info
```

```python
df.shape                                                            # get number of rows and columns
```

## **3. Filtered Dataset**

**3.1. Filter the dataset to include only schools located in Brooklyn**


```python
df_brooklyn = df[df["borough"]=="Brooklyn"]                         # Filter the DataFrame for Brooklyn
df_brooklyn.shape                                                   # New shape after filtering df to Brooklyn.
```

## **4. Answers of the following:**

**4.1. How many total schools are in Brooklyn?**
- (use the method that counts only unique values. Also there is a column that has a unique identifier for schools.)

```python
nber_schools = df_brooklyn[["dbn"]].nunique()                          # Get the number of unique schools in Brooklyn   
nber_schools
```

**4.2. How many schools in Brooklyn offer Grade 9 entry?**
- (here you need to use range to get the correct answer for example just stating min_grade entry >=9 will miss the schools that offer earlier entry))

```python
# Number of schools which offer grades spanning 9 either in minimum or maximum grade span
school_grad9_entry = df_brooklyn["dbn"][(df_brooklyn["grade_span_min"] <= 9) & (df_brooklyn["grade_span_max"] >= 9)].nunique()
school_grad9_entry
```

```python
schools_below_10 = df_brooklyn["dbn"][df_brooklyn["grade_span_max"] < 10].nunique() / df_brooklyn["dbn"].count() * 100 # Percentage of schools in Brooklyn with grade_span_max < 10
schools_above_9 = df_brooklyn["dbn"][df_brooklyn["grade_span_max"] > 9].nunique() / df_brooklyn["dbn"].count() * 100 # Percentage of schools in Brooklyn with grade_span_max > 9
print(f"Percentage of schools in Brooklyn with grade_span_max < 10: {schools_below_10:.2f}%")
print(f"Percentage of schools in Brooklyn with grade_span_max > 9: {schools_above_9:.2f}%")


# The result shows that 9 is the bigger minimum grade span for schools in Brooklyn.
# There are no schools in Brooklyn with a maximum grade span less than 10. 
# Meaning if all schools in Brooklyn offer grade 9, 99.17% of them offer a greater grade span.
```

## **6. Data Visualization**

**6.1. Bar chart: Number of schools per borough**

```python
plt.figure(figsize = (10, 6))                             # Set the figure size
plt.bar(
    nber_schools_x_borough.index,                         # Set the x-axis labels
    nber_schools_x_borough.values,                        # Set the y-axis values
    color = "skyblue"                                     # Set the bar color
)
plt.grid(axis = "y", linestyle = "--", alpha = 0.75)      # Add gridlines
plt.xlabel("Borough")                                     # Add x-label
plt.ylabel("Number of schools")                           # Add y-label
plt.title("Number of schools per borough")                # Add title
plt.xticks(rotation = 45)                                 # Rotate x-ticks
plt.show()                                                # Show the plot
```

```python
plt.figure(figsize = (12, 4))                               # Set the figure size
plt.hist(
    df["total_students"],                                   # Set the data for the histogram
    bins = 10,                                              # Set the number of bins
    color = "skyblue",                                      # Set the bar color
    edgecolor = "black"                                     # Set the edge color
)
plt.grid(axis = "y", linestyle = "--", alpha = 0.75)        # Add gridlines
plt.title("Distribution of Total Students in Schools")      # Add title
plt.xlabel("Total Students")                                # Add x-label
plt.ylabel("Frequency")                                     # Add y-label
plt.show()                                                  # Show the plot
```

```python
total_students_x_borough = df.groupby("borough")["total_students"].sum()         # Total number of students per borough
total_schools_x_borough = df.groupby("borough")["dbn"].count()                   # Total number of schools per borough
avg_students_x_school = total_students_x_borough / total_schools_x_borough       # Average number of students per school per borough

print(f"Total number of students per borough: {total_students_x_borough}\n")
print(f"\nTotal number of schools per borough: {total_schools_x_borough}\n")
print(f"\nAverage number of students per school per borough: {avg_students_x_school}")
```

```python
plt.figure(figsize=(12, 5))                                     # Set the figure size
sns.boxplot(
    x = "borough",                                              # Set the x-axis variable
    y = "total_students",                                       # Set the y-axis variable
    data = df
)
plt.title("Distribution of Students per Schools by Borough")    # Add title
plt.xlabel("Borough")                                           # Add x-label
plt.ylabel("Number of Students")                                # Add y-label
plt.xticks(rotation = 45)                                       # Rotate x-ticks
plt.grid(True, linestyle = "--", alpha = 0.5)                   # Add gridlines
plt.tight_layout()                                              # Adjust layout
plt.show()
```
